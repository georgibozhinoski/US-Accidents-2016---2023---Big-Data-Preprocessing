```python
data.drop(columns=['ID','Source', 'End_Time', 'End_Lat', 'End_Lng','Country','Timezone', 'Description', 'Weather_Timestamp', 'Airport_Code', 'Zipcode'], inplace=True)
data.sample(10)
```

```python
data.drop(columns=['Wind_Chill(F)'], inplace=True)
```

```python
outlier_summary = {}

for col in num_data:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = ((data[col] < lower) | (data[col] > upper)).sum()
    outlier_summary[col] = outliers

outlier_summary
```
