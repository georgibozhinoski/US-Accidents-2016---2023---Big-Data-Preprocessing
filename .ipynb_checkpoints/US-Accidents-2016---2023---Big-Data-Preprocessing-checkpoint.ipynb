```python
wind_map = {
    
    'N': 'North', 'North': 'North', 'NNE': 'North', 'NNW': 'North',

    
    'NE': 'Northeast', 'ENE': 'Northeast',

    
    'E': 'East', 'East': 'East', 'ESE': 'East',

    
    'SE': 'Southeast', 'SSE': 'Southeast',

   
    'S': 'South', 'South': 'South', 'SSW': 'South',

    
    'SW': 'Southwest', 'WSW': 'Southwest',

    
    'W': 'West', 'West': 'West', 'WNW': 'West',

    
    'NW': 'Northwest',

    
    'Calm': 'Calm', 'CALM': 'Calm',

    
    'Variable': 'Variable', 'VAR': 'Variable',
}

data['Wind_Group'] = data['Wind_Direction'].map(wind_map)
data['Wind_Group'] = data['Wind_Group'].fillna(data['Wind_Group'].mode()[0])
data.drop(columns=['Wind_Direction'], inplace=True)

print(data['Wind_Group'].value_counts())
```

```python
missing = data.isnull().sum()
missing = missing[missing > 0]
print(missing.sort_values(ascending=False))
print("\nAs percentage:")
print((missing / len(data) * 100).round(2))
```

```python

for col in ['Temperature(F)', 'Humidity(%)', 'Pressure(in)', 'Visibility(mi)', 'Wind_Speed(mph)']:
    data[col] = data[col].fillna(data[col].median())


data['Sunrise_Sunset'] = data['Sunrise_Sunset'].fillna(data['Sunrise_Sunset'].mode()[0])


data['City'] = data['City'].fillna('Unknown')


print(data.isnull().sum().sum())  
```

```python
import matplotlib.pyplot as plt

numerical_cols = ['Temperature(F)', 'Humidity(%)', 'Pressure(in)',
                  'Visibility(mi)', 'Wind_Speed(mph)', 'Distance(mi)']

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()

for i, col in enumerate(numerical_cols):
    axes[i].boxplot(data[col].dropna(), vert=False)
    axes[i].set_title(col)

plt.tight_layout()
plt.show()
```

```python
for col in numerical_cols:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = data[(data[col] < lower) | (data[col] > upper)].shape[0]
    print(f"{col}: lower={lower:.2f}, upper={upper:.2f}, outliers={outliers} ({outliers/len(data)*100:.2f}%)")
```

```python
def cap_outliers(df, col, lower, upper):
    df[col] = df[col].clip(lower=lower, upper=upper)


cap_outliers(data, 'Temperature(F)', lower=-50, upper=130)


cap_outliers(data, 'Pressure(in)', lower=27.0, upper=32.0)


cap_outliers(data, 'Wind_Speed(mph)', lower=0, upper=74)

print("Done. Verify:")
for col in ['Temperature(F)', 'Pressure(in)', 'Wind_Speed(mph)']:
    print(f"{col}: min={data[col].min():.2f}, max={data[col].max():.2f}")
```

```python

bool_cols = ['Amenity', 'Bump', 'Crossing', 'Give_Way', 'Junction', 'No_Exit',
             'Railway', 'Roundabout', 'Station', 'Stop', 'Traffic_Calming',
             'Traffic_Signal']

data[bool_cols] = data[bool_cols].astype(int)


from category_encoders import TargetEncoder

target_encoder = TargetEncoder(cols=['City', 'County', 'State'])
data[['City', 'County', 'State']] = target_encoder.fit_transform(
    data[['City', 'County', 'State']], data['Severity']
)


data = pd.get_dummies(data, columns=['Sunrise_Sunset', 'weather_group', 'Wind_Group'],
                      drop_first=True, dtype=int)


print(data.dtypes)
print(f"\nTotal columns: {data.shape[1]}")
```

```python
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import RandomUnderSampler
from imblearn.pipeline import Pipeline

X = data.drop(columns=['Severity'])
y = data['Severity']


undersample = RandomUnderSampler(sampling_strategy={2: 1500000}, random_state=42)
oversample = SMOTE(sampling_strategy={1: 300000, 4: 300000}, random_state=42)

pipeline = Pipeline([('under', undersample), ('over', oversample)])
X_resampled, y_resampled = pipeline.fit_resample(X, y)

# Check new distribution
print(pd.Series(y_resampled).value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler


X_train, X_test, y_train, y_test = train_test_split(
    X_resampled, y_resampled,
    test_size=0.2,
    random_state=42,
    stratify=y_resampled 
)

print(f"X_train: {X_train.shape}")
print(f"X_test:  {X_test.shape}")


scaler = StandardScaler()

numerical_cols = ['Start_Lat', 'Start_Lng', 'Distance(mi)', 'Temperature(F)',
                  'Humidity(%)', 'Pressure(in)', 'Visibility(mi)', 'Wind_Speed(mph)',
                  'City', 'County', 'State']

X_train[numerical_cols] = scaler.fit_transform(X_train[numerical_cols])
X_test[numerical_cols]  = scaler.transform(X_test[numerical_cols]) 
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import xgboost as xgb
import lightgbm as lgb
import time

results = {}


print("Training Logistic Regression...")
start = time.time()
lr = LogisticRegression(
    max_iter=1000,
    random_state=42,
    class_weight='balanced'
)
lr.fit(X_train, y_train)
lr_time = time.time() - start
y_pred_lr = lr.predict(X_test)
results['Logistic Regression'] = {
    'accuracy': accuracy_score(y_test, y_pred_lr),
    'time': lr_time,
    'report': classification_report(y_test, y_pred_lr),
    'cm': confusion_matrix(y_test, y_pred_lr),
    'preds': y_pred_lr
}
print(f"Done in {lr_time:.1f}s — Accuracy: {results['Logistic Regression']['accuracy']:.4f}\n")


print("Training Random Forest...")
start = time.time()
rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_split=50,
    min_samples_leaf=20,
    n_jobs=-1,
    random_state=42,
    class_weight='balanced'
)
rf.fit(X_train, y_train)
rf_time = time.time() - start
y_pred_rf = rf.predict(X_test)
results['Random Forest'] = {
    'accuracy': accuracy_score(y_test, y_pred_rf),
    'time': rf_time,
    'report': classification_report(y_test, y_pred_rf),
    'cm': confusion_matrix(y_test, y_pred_rf),
    'preds': y_pred_rf
}
print(f"Done in {rf_time:.1f}s — Accuracy: {results['Random Forest']['accuracy']:.4f}\n")


print("Training XGBoost...")
start = time.time()
xgb_model = xgb.XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric='mlogloss',
    device='cuda',          # GPU
    tree_method='hist'
)
# XGBoost needs labels starting from 0
xgb_model.fit(X_train, y_train - 1)
y_pred_xgb = xgb_model.predict(X_test) + 1
results['XGBoost'] = {
    'accuracy': accuracy_score(y_test, y_pred_xgb),
    'time': time.time() - start,
    'report': classification_report(y_test, y_pred_xgb),
    'cm': confusion_matrix(y_test, y_pred_xgb),
    'preds': y_pred_xgb
}
print(f"Done in {results['XGBoost']['time']:.1f}s — Accuracy: {results['XGBoost']['accuracy']:.4f}\n")


print("Training LightGBM...")
start = time.time()
lgb_model = lgb.LGBMClassifier(
    n_estimators=200,
    max_depth=10,
    learning_rate=0.1,
    num_leaves=63,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    class_weight='balanced',
    device='gpu'
)
lgb_model.fit(X_train, y_train)
lgb_time = time.time() - start
y_pred_lgb = lgb_model.predict(X_test)
results['LightGBM'] = {
    'accuracy': accuracy_score(y_test, y_pred_lgb),
    'time': lgb_time,
    'report': classification_report(y_test, y_pred_lgb),
    'cm': confusion_matrix(y_test, y_pred_lgb),
    'preds': y_pred_lgb
}
print(f"Done in {lgb_time:.1f}s — Accuracy: {results['LightGBM']['accuracy']:.4f}\n")


print("\n========== SUMMARY ==========")
summary = pd.DataFrame({
    'Model': list(results.keys()),
    'Accuracy': [v['accuracy'] for v in results.values()],
    'Training Time (s)': [round(v['time'], 1) for v in results.values()]
}).sort_values('Accuracy', ascending=False)
print(summary.to_string(index=False))


fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

for i, (name, res) in enumerate(results.items()):
    sns.heatmap(res['cm'], annot=True, fmt='d', cmap='Blues', ax=axes[i],
                xticklabels=[1,2,3,4], yticklabels=[1,2,3,4])
    axes[i].set_title(f"{name}\nAccuracy: {res['accuracy']:.4f}")
    axes[i].set_xlabel('Predicted')
    axes[i].set_ylabel('Actual')

plt.tight_layout()
plt.show()


for name, res in results.items():
    print(f"\n========== {name} ==========")
    print(res['report'])
```
